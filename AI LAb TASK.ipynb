```
# Activity 1: Basic DFS Implementation

# Graph ko adjacency list (dictionary) ke taur par define kiya gaya hai [cite: 31]
graph = {
    '5': ['3', '7'],
    '3': ['2', '4'],
    '7': ['8'],
    '2': [],
    '4': ['8'],
    '8': []
}

visited = set() # Un nodes ka record rakhne ke liye jo traverse ho chuke hain [cite: 46]

def dfs(visited, graph, node):
    if node not in visited:
        print(node, end=" ") # Node ko print karein
        visited.add(node)    # Node ko visited mark karein [cite: 50]
        for neighbour in graph[node]:
            dfs(visited, graph, neighbour) # Har padosi node ke liye recursive call [cite: 52]

print("Following is the Depth-First Search (Activity 1):")
dfs(visited, graph, '5') # Root node '5' se start karein [cite: 54]
print("\n")
```

```
# Activity 2: DFS with Cycle Handling
from collections import defaultdict

class Graph:
    def __init__(self):
        # Graph store karne ke liye default dictionary [cite: 76]
        self.graph = defaultdict(list)

    # Graph mein edge (rasta) jodne ka function [cite: 77]
    def addEdge(self, u, v):
        self.graph[u].append(v)

    # DFS ka helper function jo recursion use karta hai [cite: 80]
    def DFSUtil(self, v, visited):
        # Current node ko visited mark karein aur print karein [cite: 81, 83]
        visited[v] = True
        print(v, end=" ")

        # Saare adjacent vertices ke liye check karein [cite: 87]
        for i in self.graph[v]:
            if visited[i] == False:
                self.DFSUtil(i, visited)

    # Poore graph ko traverse karne ka main function [cite: 92]
    def DFS(self):
        # Shuruat mein sabko unvisited (False) mark karein [cite: 94]
        visited = [False] * (max(self.graph) + 1)

        # Har node ke liye check karein (agar koi disconnected part ho) [cite: 97]
        for i in range(len(visited)):
            if visited[i] == False:
                self.DFSUtil(i, visited)

# Driver Code [cite: 100]
g = Graph()
g.addEdge(0, 1)
g.addEdge(0, 2)
g.addEdge(1, 2)
g.addEdge(2, 0)
g.addEdge(2, 3)
g.addEdge(3, 3)

print("Following is Depth First Traversal (Activity 2):")
g.DFS() # Iska output '2 0 1 3' ya similar hona chahiye [cite: 63]
```

```
# Lab Task 1: DFS from Arad to Bucharest

# Romania Graph representation (City: [(Neighbor, Distance)])
romania_map = {
    'Arad': [('Zerind', 75), ('Sibiu', 140), ('Timisoara', 118)],
    'Zerind': [('Arad', 75), ('Oradea', 71)],
    'Oradea': [('Zerind', 71), ('Sibiu', 151)],
    'Sibiu': [('Arad', 140), ('Oradea', 151), ('Fagaras', 99), ('Rimnicu Vilcea', 80)],
    'Timisoara': [('Arad', 118), ('Lugoj', 111)],
    'Lugoj': [('Timisoara', 111), ('Mehadia', 70)],
    'Mehadia': [('Lugoj', 70), ('Drobeta', 75)],
    'Drobeta': [('Mehadia', 75), ('Craiova', 120)],
    'Craiova': [('Drobeta', 120), ('Rimnicu Vilcea', 146), ('Pitesti', 138)],
    'Rimnicu Vilcea': [('Sibiu', 80), ('Craiova', 146), ('Pitesti', 97)],
    'Fagaras': [('Sibiu', 99), ('Bucharest', 211)],
    'Pitesti': [('Rimnicu Vilcea', 97), ('Craiova', 138), ('Bucharest', 101)],
    'Bucharest': [('Fagaras', 211), ('Pitesti', 101), ('Giurgiu', 90), ('Urziceni', 85)],
    'Giurgiu': [('Bucharest', 90)],
    'Urziceni': [('Bucharest', 85), ('Vaslui', 142), ('Hirsova', 98)],
    'Hirsova': [('Urziceni', 98), ('Eforie', 86)],
    'Eforie': [('Hirsova', 86)],
    'Vaslui': [('Urziceni', 142), ('Iasi', 92)],
    'Iasi': [('Vaslui', 92), ('Neamt', 87)],
    'Neamt': [('Iasi', 87)]
}

def dfs_path(graph, start, goal):
    stack = [(start, [start], 0)] # (current_node, path_taken, total_distance)
    visited = set()

    while stack:
        (vertex, path, distance) = stack.pop() # LIFO behavior for DFS

        if vertex not in visited:
            if vertex == goal:
                return path, distance

            visited.add(vertex)

            # Padosi shehron ko stack mein dalna
            for neighbor, weight in reversed(graph[vertex]):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor], distance + weight))

    return None, 0

# Execution
path, total_distance = dfs_path(romania_map, 'Arad', 'Bucharest')

print("Lab Task 1 Results:")
print(f"Path: {' -> '.join(path)}")
print(f"Total Distance: {total_distance} km")
```

```
# Lab Task 2: BFS vs DFS Comparison
from collections import deque

# Romania Map Data
romania_map = {
    'Arad': ['Zerind', 'Sibiu', 'Timisoara'],
    'Zerind': ['Arad', 'Oradea'],
    'Oradea': ['Zerind', 'Sibiu'],
    'Sibiu': ['Arad', 'Oradea', 'Fagaras', 'Rimnicu Vilcea'],
    'Timisoara': ['Arad', 'Lugoj'],
    'Lugoj': ['Timisoara', 'Mehadia'],
    'Mehadia': ['Lugoj', 'Drobeta'],
    'Drobeta': ['Mehadia', 'Craiova'],
    'Craiova': ['Drobeta', 'Rimnicu Vilcea', 'Pitesti'],
    'Rimnicu Vilcea': ['Sibiu', 'Craiova', 'Pitesti'],
    'Fagaras': ['Sibiu', 'Bucharest'],
    'Pitesti': ['Rimnicu Vilcea', 'Craiova', 'Bucharest'],
    'Bucharest': ['Fagaras', 'Pitesti', 'Giurgiu', 'Urziceni'],
    'Giurgiu': ['Bucharest'],
    'Urziceni': ['Bucharest', 'Vaslui', 'Hirsova'],
    'Hirsova': ['Urziceni', 'Eforie'],
    'Eforie': ['Hirsova'],
    'Vaslui': ['Urziceni', 'Iasi'],
    'Iasi': ['Vaslui', 'Neamt'],
    'Neamt': ['Iasi']
}

def get_bfs_visited_count(start, goal):
    visited = set()
    queue = deque([start])
    count = 0
    while queue:
        node = queue.popleft()
        if node not in visited:
            count += 1
            visited.add(node)
            if node == goal: return count
            for neighbor in romania_map[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return count

def get_dfs_visited_count(start, goal):
    visited = set()
    stack = [start]
    count = 0
    while stack:
        node = stack.pop()
        if node not in visited:
            count += 1
            visited.add(node)
            if node == goal: return count
            # Neighbors ko reverse order mein stack mein dalte hain taaki sequence sahi rahe
            for neighbor in reversed(romania_map[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return count

# --- Case 1: BFS visited nodes < DFS visited nodes ---
# Goal: Sibiu (Direct neighbor of Arad)
start1, goal1 = 'Arad', 'Sibiu'
print(f"Case 1 (Goal: {goal1})")
print(f"BFS Visited Count: {get_bfs_visited_count(start1, goal1)}")
print(f"DFS Visited Count: {get_dfs_visited_count(start1, goal1)}")
print("-" * 30)

# --- Case 2: BFS visited nodes > DFS visited nodes ---
# Goal: Neamt (Deep in the graph)
start2, goal2 = 'Arad', 'Neamt'
print(f"Case 2 (Goal: {goal2})")
print(f"BFS Visited Count: {get_bfs_visited_count(start2, goal2)}")
print(f"DFS Visited Count: {get_dfs_visited_count(start2, goal2)}")
```
