```python
# Simple Hill Climbing Algorithm Example

# Heuristic values of each node
heuristic = {
    'A': 10,
    'B': 8,
    'C': 6,
    'D': 4,
    'E': 2,
    'F': 0
}

# Graph representation
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def hill_climbing(start, goal):
    current = start
    path = [current]

    while current != goal:
        neighbors = graph[current]

        if not neighbors:
            print("No better neighbor found!")
            return path

        # Select neighbor with minimum heuristic value
        next_node = min(neighbors, key=lambda x: heuristic[x])

        # Move only if heuristic improves
        if heuristic[next_node] < heuristic[current]:
            current = next_node
            path.append(current)
        else:
            print("Reached local optimum!")
            return path

    return path

# Run algorithm
start_node = 'A'
goal_node = 'F'

result = hill_climbing(start_node, goal_node)

print("Path followed:", result)
```
